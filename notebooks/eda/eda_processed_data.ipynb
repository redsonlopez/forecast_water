# Gerando relatório EDA com Pandas Profiling

```python
#Instalando e importando dependencias
#%pip install ydata-profiling
from ydata_profiling import ProfileReport
import pandas as pd 
```

```python
df = pd.read_csv("..\\..\\data\\processed\\processed_water.csv") 
df.head()
```

```python
df['DATA_VENCIMENTO'] = pd.to_datetime(df['DATA_VENCIMENTO'])
df['MATRICULA'] = df['MATRICULA'].astype('category')
```

```python
df.dtypes
```

```python
df.head()
```

```python
# Gerando relatório
profile = ProfileReport(df, title="Faturas de Agua")
```

```python
profile.to_notebook_iframe()
```

```python
profile.to_file("EDA.html")
```

# Descobrindo número ideal de Lags utilizando os metodos ACF e PACF 

```python
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
plot_acf(df['VALOR_FATURA'], lags=40, title = "Função de Autocorrelação")
plt.show
```

```python
plot_pacf(df['VALOR_FATURA'], lags=10, title = "Função de Autocorrelação Parcial")
plt.show
```
